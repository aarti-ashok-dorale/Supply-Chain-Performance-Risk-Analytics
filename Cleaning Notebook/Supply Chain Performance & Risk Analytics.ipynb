# STEP 3 — EDA + Statistical Analysis

### Descriptive Statistics

```python
numeric_cols = [
    "inventory_level",
    "equipment_availability",
    "shipping_cost",
    "supplier_reliability",
    "lead_time",
    "demand",
    "customs_clearance_time",
    "disruption_likelihood",
    "delay_probability",
    "delivery_deviation"
]

df[numeric_cols].describe().T
```

```python
# average performance, variation and extreme values
stats = pd.DataFrame({
    "Mean": df[numeric_cols].mean(),
    "Median": df[numeric_cols].median(),
    "Std_Dev": df[numeric_cols].std(),
    "Min": df[numeric_cols].min(),
    "Max": df[numeric_cols].max()
})

stats.round(2)
```

### Supplier Performance Analysis

```python
supplier_analysis = df.groupby("supplier_id").agg(
    avg_reliability=("supplier_reliability", "mean"),
    avg_lead_time=("lead_time", "mean"),
    avg_shipping_cost=("shipping_cost", "mean"),
    avg_delay_probability=("delay_probability", "mean"),
    records=("supplier_id", "count")
).reset_index()
```

### Top 10 reliable suppliers

```python
supplier_analysis.sort_values(
    "avg_reliability",
    ascending=False
).head(10)
```

### Bottom 10 Reliable Suppliers

```python
supplier_analysis.sort_values(
    "avg_reliability"
).head(10)
```

### Highest Delay Suppliers

```python
supplier_analysis.sort_values(
    "avg_delay_probability",
    ascending=False
).head(10)
```

### Highest Lead-Time Suppliers

```python
supplier_analysis.sort_values(
    "avg_lead_time",
    ascending=False
).head(10)
```

### Inventory & Demand Analysis

```python
# Product-wise inventory and demand:

inventory_analysis = df.groupby("product_id").agg(
    avg_inventory=("inventory_level", "mean"),
    avg_demand=("demand", "mean"),
    records=("product_id", "count")
).reset_index()
```

```python
# Highest Demand Products

inventory_analysis.sort_values(
    "avg_demand",
    ascending=False
).head(10)
```

```python
# Lowest Inventory Products
inventory_analysis.sort_values(
    "avg_inventory"
).head(10)
```

### 75th percentile demand and  25th percentile inventory : High Demand + Low Inventory

```python
high_demand_threshold = inventory_analysis["avg_demand"].quantile(0.75)
low_inventory_threshold = inventory_analysis["avg_inventory"].quantile(0.25)

stockout_risk = inventory_analysis[
    (inventory_analysis["avg_demand"] >= high_demand_threshold) &
    (inventory_analysis["avg_inventory"] <= low_inventory_threshold)
]

stockout_risk.head(10)
```

## Shipping Cost Analysis

```python
# Country-wise logistics performance:

country_analysis = df.groupby("supplier_country").agg(
    avg_shipping_cost=("shipping_cost", "mean"),
    avg_lead_time=("lead_time", "mean"),
    avg_delay_probability=("delay_probability", "mean"),
    avg_reliability=("supplier_reliability", "mean")
).reset_index()
```

```python
# Highest Cost Countries
country_analysis.sort_values(
    "avg_shipping_cost",
    ascending=False
).head(10)
```

```python
# Lowest Cost Countries
country_analysis.sort_values(
    "avg_shipping_cost"
).head(10)
```

## Lead Time & Delivery Analysis

```python
print("Average Lead Time:", round(df["lead_time"].mean(), 2))
print("Median Lead Time:", round(df["lead_time"].median(), 2))
print("Lead Time Std Dev:", round(df["lead_time"].std(), 2))
```

```python
# Delivery deviation:

print("Average Delivery Deviation:",
      round(df["delivery_deviation"].mean(), 2))

print("Median Delivery Deviation:",
      round(df["delivery_deviation"].median(), 2))
```

```python
# Lead Time Distribution
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df["lead_time"], bins=30)

plt.title("Lead Time Distribution")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Frequency")

plt.show()
```

### Delay Probability Analysis

```python
# High-delay threshold statistical percentile 

delay_threshold = df["delay_probability"].quantile(0.75)

print("75th Percentile Delay Probability:",
      round(delay_threshold, 2))
```

```python
# High-delay records

high_delay = df[
    df["delay_probability"] >= delay_threshold
]

print("High Delay Records:", len(high_delay))
```

```python
# Risk-wise Delay
risk_delay = df.groupby("risk_class").agg(
    avg_delay=("delay_probability", "mean"),
    avg_lead_time=("lead_time", "mean"),
    avg_reliability=("supplier_reliability", "mean")
).sort_values(
    "avg_delay",
    ascending=False
)

risk_delay
```

### Supply Chain Risk Analysis

```python
# Risk distribution

risk_distribution = df["risk_class"].value_counts()

risk_distribution
```

```python
# Percentage

risk_percentage = (
    df["risk_class"]
    .value_counts(normalize=True) * 100
).round(2)

risk_percentage
```

```python
# Risk vs Performance
risk_analysis = df.groupby("risk_class").agg(
    avg_reliability=("supplier_reliability", "mean"),
    avg_disruption=("disruption_likelihood", "mean"),
    avg_delay=("delay_probability", "mean"),
    avg_lead_time=("lead_time", "mean"),
    avg_shipping_cost=("shipping_cost", "mean")
).reset_index()

risk_analysis
```

### Route Risk Analysis

```python
route_analysis = df.groupby("route_risk").agg(
    avg_delay=("delay_probability", "mean"),
    avg_lead_time=("lead_time", "mean"),
    avg_shipping_cost=("shipping_cost", "mean")
).reset_index()

route_analysis
```

### Weather & Cargo Impact

```python
# Weather
weather_analysis = df.groupby("weather_severity").agg(
    avg_delay=("delay_probability", "mean"),
    avg_lead_time=("lead_time", "mean")
).reset_index()

weather_analysis
```

```python
# Cargo Condition
cargo_analysis = df.groupby("cargo_condition").agg(
    avg_delay=("delay_probability", "mean"),
    avg_delivery_deviation=("delivery_deviation", "mean")
).reset_index()

cargo_analysis
```

### Customs Clearance Analysis

```python
customs_analysis = df.groupby("supplier_country").agg(
    avg_customs_time=("customs_clearance_time", "mean"),
    avg_delay=("delay_probability", "mean")
).reset_index()

customs_analysis.sort_values(
    "avg_customs_time",
    ascending=False
).head(10)
```

### Correlation Analysis

```python
correlation_cols = [
    "inventory_level",
    "shipping_cost",
    "supplier_reliability",
    "lead_time",
    "demand",
    "customs_clearance_time",
    "disruption_likelihood",
    "delay_probability",
    "delivery_deviation"
]

correlation = df[correlation_cols].corr()

correlation.round(2)
```

### Heatmap

```python
import seaborn as sns

plt.figure(figsize=(11,8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Supply Chain Correlation Analysis")

plt.show()
```
